```python
# Parsear el nombre del archivo para obtener el código y nombre de la estación
station = files[2]
codigo = station.split("_")[0]
nombre = station.split("_")[1].replace(".txt", "")

print(f"Código: {codigo}, Nombre: {nombre}")
```

```python
# Leer el archivo CSV de la estación seleccionada
import pandas as pd
station_df = pd.read_table(f'/content/drive/MyDrive/senamhi_cd/{station}', header=None, sep=' ', na_values=['-999','-99.9',"-3256"]) # Sin encabezados
station_df.columns = ['Año', 'Mes', 'Dia', 'Pr','Tmax', 'Tmin']  # Asignar nombres de columnas aqui antes era clima_df.columns
# Formatear las columnas
station_df['Año'] = station_df['Año'].astype(int)
station_df
```

```python
# Precipitacion mensual
precipitacion_mensual = station_df.groupby(['Año', 'Mes'])['Pr'].sum()

# Precipitacion total anual
precipitacion_anual = station_df.groupby('Año')['Pr'].sum()

# Dia mas lluvioso
dia_mas_lluvioso = station_df.loc[station_df['Pr'].idxmax()]

print(f"Precipitación mensual:\n{precipitacion_mensual}\n")
print(f"Precipitación anual:\n{precipitacion_anual}\n")
print(f"El día más lluvioso fue el {dia_mas_lluvioso['Año']:.0f}-{dia_mas_lluvioso['Mes']:.0f}-{dia_mas_lluvioso['Dia']:.0f} con {dia_mas_lluvioso['Pr']} mm de precipitación.")
```

```python
# Exporta los resultados formateados en CSV
resumen_mensual = precipitacion_mensual.reset_index()
resumen_mensual.columns = ['Año', 'Mes', 'Precipitacion_mensual_mm']
resumen_anual = precipitacion_anual.reset_index()
resumen_anual.columns = ['Año', 'Precipitacion_anual_mm']
resumen_mensual.to_csv(f'/content/drive/MyDrive/senamhi_cd/precipitacion_mensual_{nombre}.csv', index=False)
resumen_anual.to_csv(f'/content/drive/MyDrive/senamhi_cd/precipitacion_anual_{nombre}.csv', index=False)

print("Archivos CSV exportados correctamente.")
```

### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python
# Leer los datos de las estaciones y crear un DataFrame combinado
import os
files = os.listdir('/content/drive/MyDrive/senamhi_est')  # Lista los archivos en el directorio
files
```

```python
# Usar las 5 estaciones .txt
station_files = [f for f in files if f.lower().endswith('.txt')]
station_files
```

```python
# Parsear el nombre de cada estación y crear DataFrame
import pandas as pd

estaciones = {}
df_combinado = None

for f in station_files:
    nombre_est = f.split('_', 1)[1].replace('.txt', '')

    df = pd.read_table(f'/content/drive/MyDrive/senamhi_est/{f}', header=None, sep=' ', na_values=['-999','-99.9',"-3256"])
    df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin']
    df['Año'] = df['Año'].astype(int)
    df['Mes'] = df['Mes'].astype(int)
    df['Dia'] = df['Dia'].astype(int)
    df['Fecha'] = pd.to_datetime(dict(year=df['Año'], month=df['Mes'], day=df['Dia']))

    estaciones[nombre_est] = df

    df_est = df[['Fecha', 'Pr']].rename(columns={'Pr': nombre_est})
    df_combinado = df_est if df_combinado is None else pd.merge(df_combinado, df_est, on='Fecha', how='outer')

df_combinado = df_combinado.sort_values('Fecha').reset_index(drop=True)
df_combinado
```

```python
# Promedio mensual de precipitación por estación
df_combinado['Año'] = df_combinado['Fecha'].dt.year
df_combinado['Mes'] = df_combinado['Fecha'].dt.month

nombres_estaciones = list(estaciones.keys())

promedio_mensual_estaciones = df_combinado.groupby(['Año', 'Mes'])[nombres_estaciones].mean()
promedio_mensual_estaciones
```

```python
# Mes más lluvioso en cada estación (según precipitación total del mes)
precip_total_mensual = df_combinado.groupby(['Año', 'Mes'])[nombres_estaciones].sum()

mes_mas_lluvioso = {}
for est in nombres_estaciones:
    idx_max = precip_total_mensual[est].idxmax()
    valor_max = precip_total_mensual[est].max()
    mes_mas_lluvioso[est] = {'Año': idx_max[0], 'Mes': idx_max[1], 'Precipitacion_mm': valor_max}

for est, info in mes_mas_lluvioso.items():
    print(f"Estación {est}: mes más lluvioso {info['Año']}-{info['Mes']:02d} con {info['Precipitacion_mm']:.1f} mm")
```

```python
# Exportar a CSV
promedio_mensual_estaciones.to_csv('/content/drive/MyDrive/senamhi_est/promedio_mensual_por_estacion.csv')
resumen_mes_mas_lluvioso = pd.DataFrame(mes_mas_lluvioso).T
resumen_mes_mas_lluvioso.to_csv('/content/drive/MyDrive/senamhi_est/mes_mas_lluvioso_por_estacion.csv')
print("Archivos CSV exportados correctamente.")
```
